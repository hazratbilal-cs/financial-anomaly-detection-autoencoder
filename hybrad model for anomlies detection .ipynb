# 📌 1. INTRODUCTION

Fraud detection has become one of the most critical challenges in modern financial ecosystems, including:

- Credit card transactions
- Online banking transfers
- E-commerce payments
- Digital wallets and UPI systems
- Insurance claims

The core objective of any fraud detection system is:

> **To accurately identify fraudulent transactions while keeping false positives as low as possible**, because every false alarm costs time and money for investigation teams.

---

## 🚨 1.1 Challenges in Fraud Detection

Financial fraud data presents unique difficulties:

| Challenge              | Description                                                                 | Impact on Detection                  |
|------------------------|-----------------------------------------------------------------------------|--------------------------------------|
| **Class Imbalance**    | Fraudulent cases are extremely rare (usually 0.1% – 0.8%)                   | Models become biased toward "normal" |
| **Dynamic Behavior**   | Fraudsters constantly evolve their tactics                                  | Static models become outdated quickly |
| **High Dimensionality**| Dozens or hundreds of features (amount, time, location, device, merchant, etc.) | Increased complexity and noise       |
| **Hidden Patterns**    | Fraud often mimics legitimate behavior                                      | Hard to separate without context     |
| **Concept Drift**      | Normal and fraudulent patterns change over time                             | Requires continuous model updates    |
| **Real-time Requirement** | Decisions must be made in milliseconds                                   | Limits model complexity              |

---

# ❓ 2. PROBLEM STATEMENT

Traditional single machine learning or deep learning models struggle to deliver reliable performance because:

> **A single model cannot effectively capture all types of anomalies present in financial transaction data.**

Each algorithm has its own blind spots. Relying on one technique leaves gaps that sophisticated fraudsters can exploit.

---

# 🎯 3. OBJECTIVE

To design and implement a **robust hybrid anomaly detection system** that intelligently combines:

- Deep Learning (Autoencoder)
- Unsupervised Machine Learning (Isolation Forest)
- Density-based Clustering (DBSCAN)

The goal is to achieve **higher detection accuracy**, **lower false positive rate**, and **better coverage** of different fraud patterns compared to any individual model.

---

# ❌ 4. LIMITATIONS OF SINGLE MODELS

## 🔹 4.1 DBSCAN (Density-Based Spatial Clustering of Applications with Noise)

**Working Principle:**
- Groups data points into dense clusters
- Points that do not belong to any cluster are labeled as noise/anomalies

**Strengths:**
- Does not require labeled data
- Can discover clusters of arbitrary shape
- Naturally identifies noise points

**Limitations:**
- Struggles in high-dimensional spaces (curse of dimensionality)
- Highly sensitive to `eps` and `min_samples` parameters
- Cannot detect anomalies that form their own small clusters
- Computationally expensive on large datasets

---

## 🔹 4.2 Autoencoder (Reconstruction-based Deep Learning)

**Working Principle:**
- Neural network trained to reconstruct normal transactions
- High reconstruction error → likely anomaly/fraud

**Strengths:**
- Excellent at capturing complex, non-linear patterns in data
- Works well with high-dimensional and sequential data
- Can learn rich representations of "normal" behavior

**Limitations:**
- May successfully reconstruct some fraudulent transactions (if they are not extreme)
- Requires significant training data and compute
- Reconstruction error alone can produce false positives on rare but legitimate edge cases
- Less interpretable

---

## 🔹 4.3 Isolation Forest

**Working Principle:**
- Randomly partitions the data using isolation trees
- Anomalies are isolated faster (require fewer splits) because they are different

**Strengths:**
- Very fast and scalable to large datasets
- Works well with high-dimensional data
- Few hyperparameters
- Good at detecting global outliers

**Limitations:**
- Shallow trees → limited ability to capture complex feature interactions
- May miss subtle contextual or collective anomalies
- Does not leverage the sequential/time-series nature of transactions well

---

## ⚠️ 4.4 Key Insight

> **No single model is perfect.**  
> DBSCAN misses pattern-based fraud, Autoencoders can miss extreme outliers, and Isolation Forest lacks depth for complex relationships.  
> **A hybrid approach that combines their complementary strengths is the most practical and powerful solution** for real-world financial fraud detection.

---

# 🧩 5. HYBRID MODEL CONCEPT

## 📖 5.1 Definition

A **hybrid model** (also called an ensemble or multi-model system) combines multiple heterogeneous algorithms. Each algorithm analyzes the data from a different perspective, and their outputs are intelligently fused to produce a final decision.

## 🧠 5.2 Core Concept

The fundamental idea behind this hybrid system is:

> **"Different fraud types leave different footprints in the data. By attacking the problem from multiple angles simultaneously, we cover each other's weaknesses and create a much stronger detection net."**

This hybrid system works on the principle of **multi-perspective anomaly detection**:

- **DBSCAN** → Detects **spatial / density-based** anomalies (transactions that fall outside normal dense regions)
- **Autoencoder** → Detects **pattern / reconstruction** anomalies (transactions that don't fit the learned normal manifold)
- **Isolation Forest** → Detects **isolation-based / global** outliers (transactions that are statistically rare and easy to isolate)

By running all three in parallel and combining their signals, the system becomes significantly more robust.

| **Strength**    | Detects local density deviations             |
| **Best For**    | Finding groups of suspicious behavior        |

| Aspect          | Details                                           |

## 🔹 6.3 Isolation Forest

| Aspect          | Details                                      |
|-----------------|----------------------------------------------|
| **Type**        | Unsupervised tree-based anomaly detection    |
| **Strength**    | Extremely fast and scalable                  |
| **Best For**    | Detecting global statistical outliers        |
| **Limitation**  | Limited depth and pattern understanding      |
| **Role in Hybrid** | Acts as a fast, reliable first-line filter   |

# ⚡ 7. WHY COMBINE THESE THREE MODELS?

## 🧠 7.1 Complementary Nature

| Fraud / Anomaly Type              | DBSCAN          | Autoencoder       | Isolation Forest   | Covered by Hybrid? |
|-----------------------------------|-----------------|-------------------|--------------------|--------------------|
| Extreme value outliers            | ✅ Strong       | ⚠️ Moderate       | ✅ Very Strong     | ✅ Excellent       |
| Subtle pattern deviations         | ❌ Weak         | ✅ Very Strong    | ⚠️ Moderate        | ✅ Excellent       |
| Density / cluster violations      | ✅ Very Strong  | ⚠️ Moderate       | ⚠️ Moderate        | ✅ Excellent       |
| High-dimensional anomalies        | ❌ Weak         | ✅ Strong         | ✅ Strong          | ✅ Excellent       |
| Rare but legitimate edge cases    | ⚠️ Moderate     | ⚠️ Moderate       | ✅ Strong          | ✅ Good            |

## 🎯 7.2 Multi-Perspective Detection Strategy

- **DBSCAN** provides **local density perspective**
- **Autoencoder** provides **deep pattern / manifold perspective**
- **Isolation Forest** provides **global statistical isolation perspective**

This diversity ensures that almost no type of fraud can hide from all three models at the same time.

## 🔥 7.3 Key Idea

> The hybrid model does **not** rely on any single model being correct.  
> Instead, it looks for **consensus and weighted evidence** across multiple independent detection mechanisms. This dramatically reduces both false negatives (missed fraud) and false positives (wrongly flagged transactions).

## 9.2 Step-by-Step Pipeline Description

1. **Data Ingestion** — Real-time or batch transaction stream
2. **Preprocessing** — Missing value handling, feature engineering, StandardScaler / RobustScaler
3. **Parallel Model Execution**:
   - DBSCAN clustering + density scoring
   - Autoencoder reconstruction error calculation
   - Isolation Forest anomaly score
4. **Score Normalization** — Bring all scores to comparable [0, 1] range
5. **Score Fusion** — Weighted combination (recommended) or voting
6. **Thresholding** — Apply optimized threshold
7. **Alert + Explanation** — Flag transaction + show which models triggered it (for investigators)

# 🧠 11. DECISION LOGIC & SCORE FUSION

## 11.1 Comparison of Fusion Strategies

| Strategy          | Description                                      | False Positive Risk | Fraud Coverage | Recommendation |
|-------------------|--------------------------------------------------|---------------------|----------------|----------------|
| **Simple OR**     | Flag if **any** model says fraud                 | Very High           | Highest        | Not recommended |
| **Simple AND**    | Flag only if **all** models agree                | Very Low            | Lowest         | Too conservative |
| **Weighted Score** (Recommended) | Combine normalized scores with learned weights | Low–Medium       | High           | **Best Choice** |
| **Stacking / Meta-Learner** | Train a meta-model on top of the three scores | Medium             | Very High      | Good for large labeled data |

## 11.2 Recommended Approach: Weighted Anomaly Score

```python
# Example fusion logic
normalized_dbscan     = scale_to_0_1(dbscan_outlier_score)
normalized_ae         = scale_to_0_1(autoencoder_reconstruction_error)
normalized_iforest    = scale_to_0_1(isolation_forest_score)

final_anomaly_score = (
    0.25 * normalized_dbscan +
    0.45 * normalized_ae +        # Usually give higher weight to Autoencoder
    0.30 * normalized_iforest
)

is_fraud = final_anomaly_score > optimal_threshold
```

**Why these weights?**
- Autoencoder usually captures the most nuanced patterns → higher weight
- Isolation Forest is fast and reliable for outliers → medium weight
- DBSCAN adds local density signal but is noisier in high dimensions → lower weight

The optimal weights and threshold should be tuned using a validation set with proper fraud labels (or proxy labels).

# 📊 12. EVALUATION METRICS (Critical for Fraud)

Because of severe class imbalance, **accuracy is misleading**. Use:

| Metric                    | Why It Matters                              | Target Direction |
|---------------------------|---------------------------------------------|------------------|
| **Precision**             | Of all flagged transactions, how many are real fraud? | Higher is better |
| **Recall (Sensitivity)**  | Of all actual fraud, how many did we catch? | Higher is better |
| **F1-Score**              | Balance between Precision and Recall        | Higher is better |
| **AUC-ROC**               | Overall ranking ability                     | Higher is better |
| **AUC-PR** (Precision-Recall) | More informative than AUC-ROC for imbalanced data | Higher is better |
| **False Positive Rate**   | How many normal transactions are wrongly flagged | Lower is better |
| **Alert Rate**            | Percentage of transactions sent for review  | Should be manageable for analysts |

**Business Goal**: Maximize fraud caught while keeping the number of alerts reviewable by the fraud team.

````python
#!/usr/bin/env python3
"""
================================================================================
     🧠 HYBRID FRAUD DETECTION - TERMINAL OUTPUT VERSION (No Image Files)
================================================================================
This version shows ALL results directly in the terminal / notebook output.
No files are saved in 'outputs/' folder.

Perfect for Jupyter Notebook / Google Colab / Terminal.

Just run and see everything printed + plots shown inline.
================================================================================
"""

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import IsolationForest
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler, MinMaxScaler, LabelEncoder
from sklearn.decomposition import PCA
from sklearn.metrics import (
    classification_report, confusion_matrix,
    roc_curve, auc, precision_recall_curve, average_precision_score
)
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
import warnings
warnings.filterwarnings('ignore')

# ====================== CONFIGURATION =================
DATA_PATH = "dataset.csv"
TARGET_COLUMN = "isFraud"
COLUMNS_TO_DROP = ['nameOrig', 'nameDest', 'prediction', 'isFlaggedFraud']

WEIGHT_ISOLATION = 0.30
WEIGHT_AUTOENCODER = 0.45
WEIGHT_DBSCAN = 0.25

EPOCHS = 40
BATCH_SIZE = 512
LEARNING_RATE = 0.001
# ====================================================

sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)

print("=" * 90)
print("🧠 HYBRID FRAUD DETECTION SYSTEM - TERMINAL OUTPUT VERSION")
print("=" * 90)


def load_and_prepare_data():
    print(f"\n📂 Loading dataset from: {DATA_PATH}")
    df = pd.read_csv(DATA_PATH)
    print(f"✅ Loaded successfully! Shape: {df.shape}")

    cols_to_drop = [c for c in COLUMNS_TO_DROP if c in df.columns]
    if cols_to_drop:
        df = df.drop(columns=cols_to_drop)
        print(f"🗑️  Dropped columns: {cols_to_drop}")

    if 'type' in df.columns:
        print("🔤 Encoding 'type' column...")
        le = LabelEncoder()
        df['type'] = le.fit_transform(df['type'])

    # Remove rows with missing target
    initial = len(df)
    df = df.dropna(subset=[TARGET_COLUMN])
    if len(df) < initial:
        print(f"⚠️  Dropped {initial - len(df)} rows with missing target values")

    print(f"\n📊 Final dataset shape: {df.shape}")
    fraud_rate = df[TARGET_COLUMN].mean() * 100
    print(f"📊 Fraud Rate: {fraud_rate:.3f}%")
    print(f"📊 Class Distribution:\n{df[TARGET_COLUMN].value_counts()}")

    return df


def preprocess_data(df):
    feature_cols = [col for col in df.columns if col != TARGET_COLUMN]
    print(f"\n✅ Using {len(feature_cols)} features for modeling")

    X = df[feature_cols].copy()
    y = df[TARGET_COLUMN].values.astype(int)

    X = X.fillna(X.median())

    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    split_idx = int(len(X_scaled) * 0.75)
    X_train, X_test = X_scaled[:split_idx], X_scaled[split_idx:]
    y_train, y_test = y[:split_idx], y[split_idx:]

    print(f"✅ Preprocessing complete")
    print(f"   Training samples   : {len(X_train):,}")
    print(f"   Testing samples    : {len(X_test):,}")
    print(f"   Fraud in test set  : {y_test.sum()} ({y_test.mean()*100:.3f}%)")

    return X_train, X_test, y_train, y_test, scaler, X_scaled, y, feature_cols


# ==================== MODELS ====================

class Autoencoder(nn.Module):
    def __init__(self, input_dim):
        super().__init__()
        self.encoder = nn.Sequential(nn.Linear(input_dim, 32), nn.ReLU(),
                                     nn.Linear(32, 12), nn.ReLU())
        self.decoder = nn.Sequential(nn.Linear(12, 32), nn.ReLU(),
                                     nn.Linear(32, input_dim))
    def forward(self, x):
        return self.decoder(self.encoder(x))


def train_isolation_forest(X_train):
    print("\n🌲 Training Isolation Forest...")
    model = IsolationForest(n_estimators=150, contamination=0.012, random_state=42, n_jobs=-1)
    model.fit(X_train)
    print("   ✅ Isolation Forest trained successfully")
    return model


def get_if_scores(model, X):
    return -model.decision_function(X)


def get_dbscan_scores(X, eps=0.7, min_samples=15):
    print("\n🔵 Running DBSCAN clustering...")
    db = DBSCAN(eps=eps, min_samples=min_samples, n_jobs=-1)
    labels = db.fit_predict(X)
    n_outliers = np.sum(labels == -1)
    print(f"   DBSCAN found {n_outliers} outliers ({n_outliers/len(X)*100:.2f}%)")

    from sklearn.neighbors import NearestNeighbors
    core = labels != -1
    if core.sum() > 0:
        nbrs = NearestNeighbors(n_neighbors=1).fit(X[core])
        dist, _ = nbrs.kneighbors(X)
        return dist.flatten()
    return np.ones(len(X)) * 3.0


def train_autoencoder(X_train, epochs=EPOCHS):
    print("\n🧠 Training Autoencoder (PyTorch)...")
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"   Using device: {device}")

    model = Autoencoder(X_train.shape[1]).to(device)
    criterion = nn.MSELoss()
    optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE)

    tensor = torch.FloatTensor(X_train).to(device)
    loader = DataLoader(TensorDataset(tensor), batch_size=BATCH_SIZE, shuffle=True)

    model.train()
    for epoch in range(epochs):
        total_loss = 0
        for batch in loader:
            x = batch[0]
            optimizer.zero_grad()
            loss = criterion(model(x), x)
            loss.backward()
            optimizer.step()
            total_loss += loss.item()
        if (epoch + 1) % 10 == 0:
            print(f"   Epoch [{epoch+1:2d}/{epochs}] | Loss: {total_loss/len(loader):.6f}")

    model.eval()
    with torch.no_grad():
        recon = model(tensor)
        errors = torch.mean((tensor - recon)**2, dim=1).cpu().numpy()

    print("   ✅ Autoencoder training completed")
    return model, device


def get_ae_scores(model, X, device):
    model.eval()
    t = torch.FloatTensor(X).to(device)
    with torch.no_grad():
        recon = model(t)
        return torch.mean((t - recon)**2, dim=1).cpu().numpy()


def normalize_scores(scores):
    return MinMaxScaler().fit_transform(scores.reshape(-1, 1)).flatten()


def weighted_fusion(if_scores, ae_scores, db_scores):
    if_norm = normalize_scores(if_scores)
    ae_norm = normalize_scores(ae_scores)
    db_norm = normalize_scores(db_scores)

    final = (WEIGHT_ISOLATION * if_norm +
             WEIGHT_AUTOENCODER * ae_norm +
             WEIGHT_DBSCAN * db_norm)
    return final


def find_best_threshold(y_true, scores):
    precision, recall, thresholds = precision_recall_curve(y_true, scores)
    f2_scores = (1 + 4) * (precision * recall) / (4 * precision + recall + 1e-8)
    best_idx = np.argmax(f2_scores)
    return thresholds[best_idx] if best_idx < len(thresholds) else 0.5


# ==================== VISUALIZATION (Shown in Terminal/Notebook) ====================

def show_score_distributions(if_scores, ae_scores, db_scores, final_scores, y_test):
    print("\n" + "="*70)
    print("📊 SCORE DISTRIBUTIONS (Visual)")
    print("="*70)

    fig, axes = plt.subplots(2, 2, figsize=(14, 10))

    sns.histplot(if_scores[y_test==0], bins=50, ax=axes[0,0], color='skyblue', label='Normal', stat='density')
    sns.histplot(if_scores[y_test==1], bins=30, ax=axes[0,0], color='red', label='Fraud', stat='density')
    axes[0,0].set_title("Isolation Forest Anomaly Scores"); axes[0,0].legend()

    sns.histplot(ae_scores[y_test==0], bins=50, ax=axes[0,1], color='skyblue', label='Normal', stat='density')
    sns.histplot(ae_scores[y_test==1], bins=30, ax=axes[0,1], color='red', label='Fraud', stat='density')
    axes[0,1].set_title("Autoencoder Reconstruction Error"); axes[0,1].legend()

    sns.histplot(db_scores[y_test==0], bins=50, ax=axes[1,0], color='skyblue', label='Normal', stat='density')
    sns.histplot(db_scores[y_test==1], bins=30, ax=axes[1,0], color='red', label='Fraud', stat='density')
    axes[1,0].set_title("DBSCAN Anomaly Scores"); axes[1,0].legend()

    sns.histplot(final_scores[y_test==0], bins=50, ax=axes[1,1], color='skyblue', label='Normal', stat='density')
    sns.histplot(final_scores[y_test==1], bins=30, ax=axes[1,1], color='red', label='Fraud', stat='density')
    axes[1,1].set_title("Final Weighted Fused Score"); axes[1,1].legend()

    plt.tight_layout()
    plt.show()   # Show in notebook/terminal


def show_roc_pr_curves(y_test, final_scores):
    print("\n" + "="*70)
    print("📈 ROC & PRECISION-RECALL CURVES")
    print("="*70)

    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    fpr, tpr, _ = roc_curve(y_test, final_scores)
    roc_auc = auc(fpr, tpr)
    axes[0].plot(fpr, tpr, color='darkorange', lw=2, label=f'AUC = {roc_auc:.4f}')
    axes[0].plot([0, 1], [0, 1], color='navy', lw=1, linestyle='--')
    axes[0].set_xlabel('False Positive Rate')
    axes[0].set_ylabel('True Positive Rate')
    axes[0].set_title('ROC Curve')
    axes[0].legend(loc="lower right")
    axes[0].grid(True)

    precision, recall, _ = precision_recall_curve(y_test, final_scores)
    ap = average_precision_score(y_test, final_scores)
    axes[1].plot(recall, precision, color='darkgreen', lw=2, label=f'AP = {ap:.4f}')
    axes[1].set_xlabel('Recall')
    axes[1].set_ylabel('Precision')
    axes[1].set_title('Precision-Recall Curve')
    axes[1].legend(loc="upper right")
    axes[1].grid(True)

    plt.tight_layout()
    plt.show()


def show_confusion_matrix(y_test, y_pred):
    print("\n" + "="*70)
    print("🔲 CONFUSION MATRIX")
    print("="*70)

    cm = confusion_matrix(y_test, y_pred)
    print("\nConfusion Matrix:")
    print(cm)
    print(f"\nTrue Negatives  : {cm[0,0]:,}")
    print(f"False Positives : {cm[0,1]:,}")
    print(f"False Negatives : {cm[1,0]:,}")
    print(f"True Positives  : {cm[1,1]:,}")

    plt.figure(figsize=(6, 5))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=['Normal', 'Fraud'],
                yticklabels=['Normal', 'Fraud'])
    plt.title("Confusion Matrix - Hybrid Fraud Detection")
    plt.ylabel('True Label')
    plt.xlabel('Predicted Label')
    plt.tight_layout()
    plt.show()


def show_pca_visualization(X_test, y_test, y_pred):
    print("\n" + "="*70)
    print("🔍 PCA 2D VISUALIZATION")
    print("="*70)

    pca = PCA(n_components=2)
    X_pca = pca.fit_transform(X_test)

    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    scatter1 = axes[0].scatter(X_pca[:, 0], X_pca[:, 1], c=y_test, cmap='coolwarm', alpha=0.5, s=8)
    axes[0].set_title("True Fraud Labels (PCA Projection)")
    axes[0].legend(*scatter1.legend_elements(), title="Fraud")

    scatter2 = axes[1].scatter(X_pca[:, 0], X_pca[:, 1], c=y_pred, cmap='coolwarm', alpha=0.5, s=8)
    axes[1].set_title("Hybrid Model Predictions (PCA Projection)")
    axes[1].legend(*scatter2.legend_elements(), title="Predicted")

    plt.tight_layout()
    plt.show()


def show_final_score_distribution(final_scores, y_test, threshold):
    print("\n" + "="*70)
    print("📌 FINAL WEIGHTED SCORE DISTRIBUTION + THRESHOLD")
    print("="*70)

    plt.figure(figsize=(12, 5))
    sns.histplot(final_scores[y_test==0], bins=60, color='skyblue', label='Normal', stat='density')
    sns.histplot(final_scores[y_test==1], bins=35, color='red', label='Fraud', stat='density')
    plt.axvline(x=threshold, color='darkgreen', linestyle='--', linewidth=3,
                label=f'Decision Threshold = {threshold:.4f}')
    plt.title("Final Weighted Anomaly Score Distribution")
    plt.xlabel("Fused Anomaly Score")
    plt.legend()
    plt.tight_layout()
    plt.show()


# ==================== MAIN ====================

if __name__ == "__main__":
    df = load_and_prepare_data()
    X_train, X_test, y_train, y_test, scaler, X_full, y_full, feature_names = preprocess_data(df)

    # === Train Models ===
    iso_model = train_isolation_forest(X_train)
    if_scores = get_if_scores(iso_model, X_test)

    db_scores = get_dbscan_scores(X_test)

    ae_model, device = train_autoencoder(X_train)
    ae_scores = get_ae_scores(ae_model, X_test, device)

    # === Weighted Fusion ===
    print("\n⚖️  Performing Weighted Score Fusion...")
    final_scores = weighted_fusion(if_scores, ae_scores, db_scores)

    # === Threshold ===
    best_threshold = find_best_threshold(y_test, final_scores)
    print(f"\n🎯 Optimal Decision Threshold (max F2-score): {best_threshold:.4f}")

    y_pred = (final_scores >= best_threshold).astype(int)

    # === Evaluation ===
    print("\n" + "="*90)
    print("📊 FINAL EVALUATION RESULTS (Your Dataset)")
    print("="*90)
    print(classification_report(y_test, y_pred, target_names=['Normal', 'Fraud']))

    # === Visualizations (shown in terminal/notebook) ===
    show_score_distributions(if_scores, ae_scores, db_scores, final_scores, y_test)
    show_roc_pr_curves(y_test, final_scores)
    show_confusion_matrix(y_test, y_pred)
    show_pca_visualization(X_test, y_test, y_pred)
    show_final_score_distribution(final_scores, y_test, best_threshold)

    print("\n" + "="*90)
    print("✅ ALL RESULTS DISPLAYED SUCCESSFULLY IN TERMINAL / NOTEBOOK")
    print("="*90)
````
